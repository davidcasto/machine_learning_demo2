```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Mary"],
    "Age": [20,22,21],
    "Marks":[80, 65, 90]
}

students = pd.DataFrame(data)

print(students)

print("==========")
print(students["Name"])
print("===========")
print(students[["Name","Marks"]])
print("====Selecting a row=====")
print(students.iloc[0])
print("====get basic information====")
print(students.shape)

print(students.columns)

print(students.dtypes)

print(students.describe())

```

```python
import pandas as pd

data = {
    "Name": ["David", "John", "Mary", "Peter", "Anna"],
    "Age": [20, 22, 21, 23, 19],
    "Marks": [80, 65, 90, 55, 75]
}

students = pd.DataFrame(data)

print("=== Complete Dataset ===")
print(students)

print("\n=== Names ===")
print(students["Name"])

print("\n=== Marks ===")
print(students["Marks"])

print("\n=== First Student ===")
print(students.iloc[0])

print("\n=== Dataset Shape ===")
print(students.shape)

print("\n=== Data Types ===")
print(students.dtypes)

print("\n=== Statistics ===")
print(students.describe())
```
